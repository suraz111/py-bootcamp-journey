Sets


sets are the built in datatype in python used to store collection of unique items. they are unordered, meaining that the element do not follow a specific order, and they donot allow the duplicate elements. Sets are useful for membership tests,eliminating duplicate entires and performing mathematical set operation like as union,intersection,difference.

```python
set={1,2,3,4}
print(type(set))
print(set)
```

```python
#set doesn't allow duplicate vaues
my_set={1,2,3,4,4,5,5,6,7,1}
print(my_set)
```

```python
##basic set operations
my_set={1,2,3,4,5}
my_set.add(6)
print(my_set)
my_set.remove(1)
print(my_set)
```

```python
my_set.discard(10) #discard doesn't throw error if the value is not present in the set  
print(my_set)
```

```python
my_set={1,2,3,4,5}
print(my_set.pop()) #pop removes a random value from the set
print(my_set)
```

```python
#to clear the set
my_set={1,2,3,4,5}
print(my_set.clear())
```

```python
##set membership
my_set={1,2,3,4,5}
print(1 in my_set)
print(10 in my_set)
```

```python
##mathematical set operations
set1={1,2,3,4,5}
set2={4,5,6,7,8}
#union of two sets
union_set=set1.union(set2)
print(union_set)
#intersection of two sets
intersection_set=set2.intersection(set1)
print(intersection_set)
#difference of two sets
difference_set=set1.difference(set2)
print(difference_set)
#symmetric difference of two sets
symmetric_difference_set=set1.symmetric_difference(set2)
print(symmetric_difference_set)
```

```python
## set methods
set1={1,2,3,4,5}
set2={4,5,6,7,8}

## is the sub set
print(set1.issubset(set2))
#superset
print(set1.issuperset(set2))
```

```python
list1=[1,2,3,3,4,5,4,5]
set_from_list =set(list1)
print(set_from_list)
```

```python
##counting the unique words in text
text="This is a sample text with some sample words. This text is just a sample."
unique_words = {word for word in text.split()}
print(unique_words)
print(len(unique_words))
```
