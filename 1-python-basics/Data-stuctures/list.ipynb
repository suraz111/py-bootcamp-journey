```python
name=['suraj','hari','sita','ram']
print(name)
name.insert(0,'anuka') #insert the elemet at given index
print(name.count('anuka'))
name.append('suraj') #insert elememt at the end of the list
print(name.count('suraj')) #count the number of times the element is present
print(name,name.sort())
name.clear() #clears the list
print(name)
```

```python
###Slicing the list
name=['suraj','hari','sita','ram']
print(name[0:3])
print(name[:4])
print(name[0:])
print(name[0::2]) #jumps the element with given steps
print(name[-1::-1]) #reverse the list
print(name[::-1]) ## reverse the list
```

```python
#iterating over the list
name=['suraj','hari','sita','ram']
for i in name:
    print(i,end=",")
```

List comprehension

Basics syntax[expression for item in iterable]
with conditional logic[expression for item in iterable if condition]

```python
### lsit comprehension
squres=[num**2 for num in range(11)]
print(squres)
```

```python
#list comphrnsion with the comdition
list=[]
for i in range(11):
    if i%2==0:
         list.append(i)
print(list)
```

```python
even=[num for num in range(11) if num%2==0]
print(even)
```

```python
#nested list comphension
list1=[1,2,3,4,5]
list2=['suraj','ramu','sita','kalu','anuka']
pair=[[i,j] for i in list1 for j in list2]
print(pair)
```
