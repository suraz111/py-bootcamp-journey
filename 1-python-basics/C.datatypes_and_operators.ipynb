```python
##arithemetic operator
a=100
b=200
sum=a+b
print("sum of a and b is:",sum)
```

```python
#subraction
a=200
b=100
sub=a-b
print("subracton of a and b is:",sub)
multiplication=a*b
print("multiplication of a and b is:",multiplication)
division=a/b
print("division of a and b is:",division)
floor_division=a//b
print("float division of a and b is:",float_division)
modulus=a%b ##modulus operator
print("modulus of a and b is:",modulus)
exponentiation=a**b
print("exponentiation of a and b is:",exponentiation)
```
