#type conversion and casting

```python
#type conversion
age=19
print(age)
print(type(age))
age_str=str(age)
print(age_str)
print(type(age_str))
```

Dynamic typing
-changing the type of variable after each iteration

```python
var="suraj"
print(var,type(var))
var=100
print(var,type(var))
var=6.1
print(var,type(var))
```
