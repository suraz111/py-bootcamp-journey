```python
for i in range(5):
    print(i)
```

while loop
continues to execute as long as the condition is True

```python
count=0
while count<10:
    print(count)
    count+=1
```

```python
#loop control statements: 
# break- it exits the loop
# continue- it skips the current iteration and continues with the next iteration
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
#continue
for i in range(10):
    if i==5:
        continue
    print(i)
```

```python
#pass
for i in range(5):
    if i==2:
        print("pass the statement")
        pass
    print(i)
```

```python
for i in range(10):
    for j in range(9):
        print(i,j)
```

```python
#calculate the sum of first N natural number
n=10
sum=0
count=1
while count<=n:
    sum=sum+count
    count=count+1
print("sum of the first 10 natual number is:",sum)
```

```python
#prime number between 1 to 100
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if (num%i)==0:
                break
        else:
                print(num,end=",")
```
